```python
import numpy as np

# Small dataset (AND gate)
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([[0],
              [0],
              [0],
              [1]])

# Sigmoid activation
def sigmoid(x):
    return 1/(1+np.exp(-x))

# Binary Cross Entropy Loss
def loss(y_true, y_pred):
    e = 1e-10
    y_pred = np.clip(y_pred, e, 1-e)
    return -np.mean(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))
```

## Single layer perceptron


```python
# Weights and Bias
W = np.array([[1],
              [1]])
b = -1.5

# Forward Propagation
z = np.dot(X, W) + b
output = sigmoid(z)

print("Output:")
print(output)

# Loss
l = loss(y, output)
print("Loss =", l)
```

## two layer feedforward perceptron

```python
# Hidden Layer
W1 = np.array([[1,1],
               [1,1]])
b1 = np.array([[0,0]])

# Output Layer
W2 = np.array([[1],
               [1]])
b2 = -1

# Forward Propagation
z1 = np.dot(X, W1) + b1
a1 = sigmoid(z1)

z2 = np.dot(a1, W2) + b2
final_output = sigmoid(z2)

print("Final Output:")
print(final_output)

```

## loss calculation

```python
# Loss
l2 = loss(y, final_output)
print("Loss =", l2)

```
